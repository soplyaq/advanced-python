```python
import matplotlib.pyplot as plt
import pandas as pd

data = {
    'Name': ['Topf', 'Katara', 'Aang', 'Zuko', 'Soka'],
    'Age': [19, 21, 20, 22, 18],
    'Score': [85, 92, 78, 95, 88]
}
df = pd.DataFrame(data)

plt.figure(figsize=(10, 6))
plt.plot(df['Name'], df['Score'], marker='o', linestyle='-', color='b')

plt.xlabel('Имя студента')
plt.ylabel('Балл')
plt.title('Баллы студентов')
plt.grid(True, alpha=0.3)

plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.bar(df['Name'], df['Score'], color='skyblue')

plt.xlabel('Students Name')
plt.ylabel('Score')
plt.title('Students Scores bar chart')

for i, score in enumerate(df['Score']):
    plt.text(i, score + 1, str(score), ha='center')

plt.show()
```
